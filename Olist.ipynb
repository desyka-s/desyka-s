# **5. Explanatory Data Analysis**

Untuk memahami kesehatan finansial bisnis e-commerce, kita akan menghitung total pendapatan kotor (GMV), jumlah transaksi, AOV, dan rata-rata waktu antar pesanan dari 2016 hingga 2018 untuk mengevaluasi pertumbuhan dan efisiensi penjualan.

```python
# Menghitung GMV
query = """
SELECT SUM(price) FROM olist;
"""

review_payments = duckdb.sql(query).to_df()
# Mengakses nilai di baris pertama dan kolom pertama (dengan asumsi itu adalah jumlahnya)
total_revenue = review_payments.iloc[0, 0]
print(f"{total_revenue:,.0f}") # Terapkan pemformatan ke nilai yang diekstraksi
```

```python
# Menghitung jumlah transaksi (bukan jumlah pelanggan, karena 1 pelanggan bisa beli berkali-kali)
query = """
SELECT COUNT(order_id) FROM olist;
"""

review_orders = duckdb.sql(query).to_df()
# Mengakses nilai di baris pertama dan kolom pertama (dengan asumsi itu adalah jumlahnya)
total_orders = review_orders.iloc[0, 0]
print(f"{total_orders:,.0f}") # Terapkan pemformatan ke nilai yang diekstraksi
```

```python
# Menghitung jumlah transaksi unik
query = """
SELECT COUNT(DISTINCT order_id) FROM olist;
"""

unique_orders_df = duckdb.sql(query).to_df()
unique_orders = unique_orders_df.iloc[0, 0]
print(f"Jumlah transaksi unik:{unique_orders:,.0f}")

# Menghitung AOV
aov = total_revenue / unique_orders

print(f"Average Order Value (AOV): {aov:,.2f}")
```

```python
# Menghitung rata-rata waktu antar pesanan
# Repeat Order Time Analysis

# Same-day repeat
query = """
CREATE OR REPLACE TABLE repeat AS
  SELECT
    customer_unique_id,
    order_purchase_timestamp::TIMESTAMP AS order_purchase,
    LEAD(order_purchase_timestamp::TIMESTAMP) OVER (
        PARTITION BY customer_unique_id
        ORDER BY order_purchase_timestamp
    ) AS next_order_date
  FROM olist;
"""
duckdb.sql(query)

# Hitung rata-rata hari antar pesanan (hanya untuk customer dengan ≥2 order)
query = """
SELECT
  AVG(DATEDIFF('day', order_purchase, next_order_date)) AS avg_days_between_orders,
  MEDIAN(DATEDIFF('day', order_purchase, next_order_date)) AS median_days_between_orders
FROM repeat
WHERE next_order_date IS NOT NULL;
"""
review_orders = duckdb.sql(query).to_df()
review_orders
```

```python
# Summary repeat order
query = """
SELECT
    COUNT(*) AS total_repeat_pairs,
    COUNT(*) FILTER (WHERE days_between = 0) AS same_day_repeat,
    100.0 * COUNT(*) FILTER (WHERE days_between = 0) / COUNT(*) AS pct_same_day,
    COUNT(*) FILTER (WHERE days_between > 0) AS real_repeat,
    100.0 * COUNT(*) FILTER (WHERE days_between > 0) / COUNT(*) AS pct_real_repeat,
    AVG(days_between) FILTER (WHERE days_between > 0) AS avg_real_repeat_days,
    MEDIAN(days_between) FILTER (WHERE days_between > 0) AS median_real_repeat_days
FROM (
    SELECT
        customer_unique_id,
        DATEDIFF('day', order_purchase, next_order_date) AS days_between
    FROM repeat
    WHERE next_order_date IS NOT NULL
);
"""
duckdb.sql(query).to_df()
```

Untuk menganalisis perilaku dan retensi pelanggan, kita akan melakukan cohort analysis terhadap data pelanggan selama 2016 hingga 2018 untuk mengukur tingkat retensi kuartal dan tren pembelian berulang.

```python
# Menghitung cohort analysis
# 1. Hitung pembelian pertama per kuartal (bukan per bulan)
duckdb.sql("""
CREATE OR REPLACE TABLE first_purchases AS
    SELECT
        c.customer_unique_id,
        MIN(DATE_TRUNC('quarter', o.order_purchase_timestamp::TIMESTAMP)) AS cohort_quarter
    FROM olist o
    JOIN customers c ON o.customer_id = c.customer_id
    WHERE o.order_status = 'delivered'
    GROUP BY c.customer_unique_id
    HAVING COUNT(DISTINCT o.order_id) > 1  -- Hanya pelanggan dengan >1 order
""")

# 2. Hitung ukuran cohort per kuartal
duckdb.sql("""
CREATE OR REPLACE TABLE cohort_size AS
    SELECT
        cohort_quarter,
        COUNT(DISTINCT customer_unique_id) AS total_customers
    FROM first_purchases
    GROUP BY cohort_quarter
""")

# 3. Hitung retention per kuartal
duckdb.sql("""
CREATE OR REPLACE TABLE retention_data AS
    SELECT
        fp.cohort_quarter,
        EXTRACT(QUARTER FROM AGE(DATE_TRUNC('quarter', o.order_purchase_timestamp::TIMESTAMP), fp.cohort_quarter)) AS quarter_number,
        COUNT(DISTINCT fp.customer_unique_id) AS retained_customers
    FROM olist o
    JOIN customers c ON o.customer_id = c.customer_id
    JOIN first_purchases fp ON c.customer_unique_id = fp.customer_unique_id
    WHERE o.order_status = 'delivered'
    GROUP BY fp.cohort_quarter, quarter_number
""")

# 4. Gabungkan hasil untuk hitung persentase retention
cohort_query_quarterly = """
SELECT
    r.cohort_quarter,
    r.quarter_number,
    r.retained_customers,
    cs.total_customers,
    (r.retained_customers::FLOAT / cs.total_customers) * 100 AS retention_rate
FROM retention_data r
JOIN cohort_size cs ON r.cohort_quarter = cs.cohort_quarter
ORDER BY r.cohort_quarter, r.quarter_number
"""

cohort_data_quarterly = duckdb.sql(cohort_query_quarterly).df()

# 5. Visualisasi Heatmap (Per Kuartal)
import seaborn as sns
import matplotlib.pyplot as plt

# Pivot data untuk heatmap
retention_matrix_quarterly = cohort_data_quarterly.pivot_table(
    index='cohort_quarter',
    columns='quarter_number',
    values='retention_rate'
)

plt.figure(figsize=(12, 8))
sns.heatmap(
    retention_matrix_quarterly,
    annot=True,
    fmt='.1f',
    cmap='YlGnBu',
    vmin=0,
    vmax=100,
    linewidths=0.5
)
plt.title('Customer Retention Rate by Quarter (%)', pad=20)
plt.xlabel('Quarters After First Purchase')
plt.ylabel('Cohort Quarter')
plt.show()
```

Untuk mengevaluasi kinerja operasional dan mitra (penjual), kita akan menilai tren penjualan musiman dengan membandingkan volume penjualan bulanan dari 2017 hingga 2018 untuk mengidentifikasi pola temporal.

```python
# Menghitung tren penjualan musiman
import matplotlib.dates as mdates
from datetime import datetime

# Query for daily order counts
daily_orders_query = """
SELECT
    order_purchase_timestamp::DATE AS order_date,
    COUNT(DISTINCT order_id) AS order_count
FROM olist
WHERE
    order_purchase_timestamp BETWEEN '2017-09-16' AND '2018-01-30'
    AND order_status = 'delivered'
GROUP BY order_date
ORDER BY order_date
"""

daily_orders = duckdb.sql(daily_orders_query).df()
daily_orders['order_date'] = pd.to_datetime(daily_orders['order_date'])

# Visualization
plt.figure(figsize=(16, 8))

# Plot with all dates shown
plt.plot(
    daily_orders['order_date'],
    daily_orders['order_count'],
    marker='o',
    markersize=6,
    color='#2ecc71',  # Emerald green
    linewidth=2.5,
    markeredgecolor='#27ae60',
    markerfacecolor='white'
)

# Configure x-axis
ax = plt.gca()
ax.xaxis.set_major_locator(mdates.DayLocator(interval=7))  # Show every 7 days
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))
plt.xticks(rotation=45, ha='right')

# Expand y-axis range (add 10% buffer)
y_max = daily_orders['order_count'].max()
plt.ylim([0, y_max * 1.3])  # 10% headroom

# Styling
plt.title('Daily Order Counts (16 Sep 2017 - 30 Jan 2018)', fontsize=18, pad=20)
plt.xlabel('Date', fontsize=14, labelpad=10)
plt.ylabel('Number of Orders', fontsize=14, labelpad=10)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Highlight important points
max_order = daily_orders['order_count'].max()
max_date = daily_orders.loc[daily_orders['order_count'].idxmax(), 'order_date']
plt.annotate(
    f'Peak: {max_order} orders\n{max_date.strftime("%b %d")}',
    xy=(max_date, max_order),
    xytext=(15, 15),
    textcoords='offset points',
    bbox=dict(boxstyle='round,pad=0.5', fc='#f39c12', alpha=0.9),
    fontsize=12,
    arrowprops=dict(arrowstyle='->', color='#34495e')
)

# Add minor ticks for better readability
ax.xaxis.set_minor_locator(mdates.DayLocator(interval=1))
ax.grid(which='minor', alpha=0.2)

plt.tight_layout()
plt.show()
```
